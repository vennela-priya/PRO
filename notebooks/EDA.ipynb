# Exploratory Data Analysis
Figshare Brain Tumor Dataset — 3064 T1-weighted MRI slices


```python
import sys; sys.path.insert(0, '..')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 120
```

```python
# Load metadata (run preprocessing first)
metadata_path = Path('../data/processed/metadata.csv')
if metadata_path.exists():
    df = pd.read_csv(metadata_path)
    print(f'Total samples: {len(df)}')
    print(df['class_name'].value_counts())
else:
    print('Run train.py first to generate processed metadata.')
    df = pd.DataFrame({'class_name': ['meningioma']*708 + ['glioma']*1426 + ['pituitary']*930,
                       'label': [0]*708 + [1]*1426 + [2]*930})
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Class distribution
counts = df['class_name'].value_counts()
colors = ['#2196F3', '#4CAF50', '#FF5722']
axes[0].bar(counts.index, counts.values, color=colors, alpha=0.85)
for i, (name, val) in enumerate(counts.items()):
    axes[0].text(i, val + 10, str(val), ha='center', fontweight='bold')
axes[0].set_title('Class Distribution', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Count')

# Pie chart
axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%',
            colors=colors, startangle=90)
axes[1].set_title('Class Proportions', fontsize=13, fontweight='bold')

plt.suptitle('Figshare Brain Tumor Dataset (n=3064)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('../results/eda_class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print('Class imbalance ratio (max/min):', counts.max() / counts.min())
```
